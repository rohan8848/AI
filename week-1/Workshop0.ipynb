```python
# Task-1

def meters_to_feet(meters):
    """Converts meters to feet."""
    return meters * 3.28084

def feet_to_meters(feet):
    """Converts feet to meters."""
    return feet / 3.28084

def kg_to_lbs(kg):
    """Converts kilograms to pounds."""
    return kg * 2.20462

def lbs_to_kg(lbs):
    """Converts pounds to kilograms."""
    return lbs / 2.20462

def liters_to_gallons(liters):
    """Converts liters to gallons."""
    return liters * 0.264172

def gallons_to_liters(gallons):
    """Converts gallons to liters."""
    return gallons / 0.264172

def get_conversion(choice, value):
    """Performs the selected conversion."""
    conversions = {
        '1': meters_to_feet,
        '2': feet_to_meters,
        '3': kg_to_lbs,
        '4': lbs_to_kg,
        '5': liters_to_gallons,
        '6': gallons_to_liters
    }
    
    if choice in conversions:
        return conversions[choice](value)
    else:
        return None

def main():
    """Main function to handle user input and perform conversions."""
    print("Unit Converter: Choose a conversion type")
    print("1: Meters to Feet")
    print("2: Feet to Meters")
    print("3: Kilograms to Pounds")
    print("4: Pounds to Kilograms")
    print("5: Liters to Gallons")
    print("6: Gallons to Liters")
    
    choice = input("Enter your choice (1-6): ")
    
    try:
        value = float(input("Enter the value to convert: "))
        result = get_conversion(choice, value)
        
        if result is not None:
            print(f"Converted value: {result:.4f}")
        else:
            print("Invalid choice. Please select a number between 1 and 6.")
    except ValueError:
        print("Invalid input. Please enter a numeric value.")

if __name__ == "__main__":
    main()
```

```python
def calculate_sum(numbers):
    """Returns the sum of the list of numbers."""
    return sum(numbers)

def calculate_average(numbers):
    """Returns the average of the list of numbers."""
    return sum(numbers) / len(numbers) if numbers else 0

def find_maximum(numbers):
    """Returns the maximum number from the list."""
    return max(numbers) if numbers else None

def find_minimum(numbers):
    """Returns the minimum number from the list."""
    return min(numbers) if numbers else None

def get_numbers():
    """Prompts the user to enter a list of numbers and returns them as a list."""
    try:
        user_input = input("Enter a list of numbers separated by spaces: ")
        numbers = list(map(float, user_input.split()))
        if not numbers:
            raise ValueError("The list cannot be empty.")
        return numbers
    except ValueError:
        print("Invalid input. Please enter numeric values only.")
        return None

def main():
    """Main function to handle user input and perform mathematical operations."""
    print("Mathematical Operations:")
    print("1: Sum")
    print("2: Average")
    print("3: Maximum")
    print("4: Minimum")
    
    choice = input("Enter your choice (1-4): ")
    numbers = get_numbers()
    
    if numbers is None:
        return
    
    operations = {
        '1': calculate_sum,
        '2': calculate_average,
        '3': find_maximum,
        '4': find_minimum
    }
    
    if choice in operations:
        result = operations[choice](numbers)
        print(f"Result: {result}")
    else:
        print("Invalid choice. Please select a number between 1 and 4.")

if __name__ == "__main__":
    main()
```

```python
def extract_every_other(lst):
    """
    Extracts every other element from a given list, starting with the first element.

    Parameters:
        lst (list): The list from which elements will be extracted.

    Returns:
        list: A new list containing every other element from the original list.
    """
    # Take every second element from the list, starting from the first one
    result = lst[::2]
    return result

# Example usage
example_list = [1, 2, 3, 4, 5, 6]
filtered_list = extract_every_other(example_list)

# Display the result
print("Original list:", example_list)
print("Every other element:", filtered_list)
```

```python
def get_sublist(lst, start, end):
    """
    Returns a sublist from the given list, starting from a specified index and ending at another specified index.

    Parameters:
        lst (list): The input list.
        start (int): The starting index (inclusive).
        end (int): The ending index (inclusive).

    Returns:
        list: A sublist from start to end (inclusive).
    """
    # Extract the sublist from start to end (inclusive)
    return lst[start:end+1]

# Example usage
example_list = [1, 2, 3, 4, 5, 6]
start_index = 2
end_index = 4

sublist = get_sublist(example_list, start_index, end_index)

# Display the result
print("Original list:", example_list)
print(f"Sublist from index {start_index} to {end_index}:", sublist)
```

```python
def reverse_list(lst):
    """
    Reverses a list using slicing.

    Parameters:
        lst (list): The input list.

    Returns:
        list: A new list with elements in reverse order.
    """
    return lst[::-1]  # Slicing with step -1 reverses the list

# Example usage
example_list = [1, 2, 3, 4, 5]
reversed_list = reverse_list(example_list)

# Display the result
print("Original list:", example_list)
print("Reversed list:", reversed_list)
```

```python
def flatten(lst):
    """
    Flattens a nested list into a single list with all elements in a single dimension.

    Parameters:
        lst (list): The nested list.

    Returns:
        list: A flattened version of the list.
    """
    flat_list = []
    for sublist in lst:
        if isinstance(sublist, list):
            flat_list.extend(sublist)  # Extend the flat list with elements from the sublist
        else:
            flat_list.append(sublist)  # Append non-list elements directly
    return flat_list

# Example usage
example_nested_list = [[1, 2], [3, 4], [5]]
flattened_list = flatten(example_nested_list)

# Display the result
print("Original nested list:", example_nested_list)
print("Flattened list:", flattened_list)
```

```python
3. Sum of All Elements in a Nested List:
```

```python
def sum_nested(lst):
    """
    Calculates the sum of all numbers in a nested list, regardless of depth.

    Parameters:
        lst (list): The nested list containing numbers.

    Returns:
        int: The sum of all numbers in the nested list.
    """
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)  # Recursively sum elements in sublists
        else:
            total += item  # Add non-list elements directly
    return total

# Example usage
example_nested_list = [[1, 2], [3, [4, 5]], 6]
nested_sum = sum_nested(example_nested_list)

# Display the result
print("Original nested list:", example_nested_list)
print("Sum of all elements:", nested_sum)
```

```python
4. Remove Specific Element from a Nested List:
```

```python
def remove_element(lst, elem):
    """
    Removes all occurrences of a specific element from a nested list.

    Parameters:
        lst (list): The nested list.
        elem: The element to be removed.

    Returns:
        list: A modified list with the specified element removed.
    """
    modified_list = []
    
    for item in lst:
        if isinstance(item, list):
            # Recursively remove element from sublists
            modified_sublist = remove_element(item, elem)
            if modified_sublist:  # Only add non-empty lists
                modified_list.append(modified_sublist)
        elif item != elem:
            modified_list.append(item)  # Add non-matching elements

    return modified_list

# Example usage
example_nested_list = [[1, 2], [3, 2], [4, 5]]
element_to_remove = 2
modified_list = remove_element(example_nested_list, element_to_remove)

# Display the result
print("Original nested list:", example_nested_list)
print(f"List after removing {element_to_remove}:", modified_list)
```

```python
5. Find the Maximum Element in a Nested List:
```

```python
def find_max(lst):
    """
    Finds the maximum element in a nested list, regardless of depth.

    Parameters:
        lst (list): The nested list.

    Returns:
        int/float: The maximum element in the nested list.
    """
    max_value = float('-inf')  # Initialize with the smallest possible number

    for item in lst:
        if isinstance(item, list):
            max_value = max(max_value, find_max(item))  # Recursively find max in sublists
        else:
            max_value = max(max_value, item)  # Compare non-list elements

    return max_value

# Example usage
example_nested_list = [[1, 2], [3, [4, 5]], 6]
max_element = find_max(example_nested_list)

# Display the result
print("Original nested list:", example_nested_list)
print("Maximum element:", max_element)
```

```python
6. Count Occurrences of an Element in a Nested List:
```

```python
def count_occurrences(lst, elem):
    """
    Counts how many times a specific element appears in a nested list.

    Parameters:
        lst (list): The nested list.
        elem: The element to count.

    Returns:
        int: The count of occurrences of the specified element.
    """
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)  # Recursively count in sublists
        elif item == elem:
            count += 1  # Increment count if element matches
    return count

# Example usage
example_nested_list = [[1, 2], [2, 3], [2, 4]]
element_to_count = 2
occurrences = count_occurrences(example_nested_list, element_to_count)

# Display the result
print("Original nested list:", example_nested_list)
print(f"Occurrences of {element_to_count}:", occurrences)
```

```python
7. Flatten a List of Lists of Lists:
```

```python
def deep_flatten(lst):
    """
    Flattens a deeply nested list into a single list, regardless of depth.

    Parameters:
        lst (list): The deeply nested list.

    Returns:
        list: A flattened version of the list.
    """
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            flat_list.extend(deep_flatten(item))  # Recursively flatten sublists
        else:
            flat_list.append(item)  # Append non-list elements directly
    return flat_list

# Example usage
example_deep_nested_list = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
flattened_list = deep_flatten(example_deep_nested_list)

# Display the result
print("Original deeply nested list:", example_deep_nested_list)
print("Flattened list:", flattened_list)
```

```python
8. Nested List Average:
```

```python
def deep_flatten(lst):
    """
    Flattens a deeply nested list into a single list, regardless of depth.

    Parameters:
        lst (list): The deeply nested list.

    Returns:
        list: A flattened version of the list.
    """
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            flat_list.extend(deep_flatten(item))  # Recursively flatten sublists
        else:
            flat_list.append(item)  # Append non-list elements directly
    return flat_list

def average_nested(lst):
    """
    Calculates the average of all elements in a nested list.

    Parameters:
        lst (list): The nested list.

    Returns:
        float: The average of all elements in the nested list.
    """
    flat_list = deep_flatten(lst)
    return sum(flat_list) / len(flat_list) if flat_list else 0

# Example usage
example_nested_list = [[1, 2], [3, 4], [5, 6]]
nested_average = average_nested(example_nested_list)

# Display the result
print("Original nested list:", example_nested_list)
print("Average of all elements:", nested_average)
```

```python
import time

# Define the size of the lists/arrays
size = 1_000_000

# Generate two lists of size 1,000,000 with random values
list1 = [i for i in range(size)]
list2 = [i for i in range(size)]

# Measure time for element-wise addition using Python lists
start_time = time.time()
list_sum = [list1[i] + list2[i] for i in range(size)]
end_time = time.time()
list_time = end_time - start_time

# Generate two numpy arrays of size 1,000,000 with the same values
array1 = np.arange(size)
array2 = np.arange(size)

# Measure time for element-wise addition using NumPy arrays
start_time = time.time()
array_sum = array1 + array2  # Vectorized operation
end_time = time.time()
numpy_time = end_time - start_time

# Display the results
list_time, numpy_time
```

```python
Element-wise Addition: Python Lists vs. NumPy Arrays
Time Taken:
Using Python Lists: 0.2503 seconds
Using NumPy Arrays: 0.0026 seconds
Observation:
NumPy arrays are nearly 95 times faster than Python lists for element-wise addition.
The primary reason for this is vectorization, where NumPy processes the entire array in one go instead of looping over elements one by one like Python lists.
NumPy also benefits from optimized memory management and low-level C implementations, making it significantly more efficient.
```
